 **K-Means Clustering**
--> What: Partitions data into K clusters (you choose K)

--> Goal: Minimize distance within clusters and maximize it between clusters

--> Needs you to predefine number of clusters (K)

---> It assigns points, recalculates centroids, and repeats until convergence

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3)
kmeans.fit(X)
labels = kmeans.labels_
```

**Hierarchical Clustering**
--> What: Builds a tree (dendrogram) of clusters

--> No need to define K at start (you can cut the tree later)

Types:
-------

Agglomerative (bottom-up) — more common

Divisive (top-down)

```python
from sklearn.cluster import AgglomerativeClustering

hier = AgglomerativeClustering(n_clusters=3)
labels = hier.fit_predict(X)
```

**DBSCAN (Density-Based Spatial Clustering)**

 What: Groups points that are closely packed together

Doesn’t require number of clusters

Can identify noise/outliers

Great for non-spherical shapes

```python
from sklearn.cluster import DBSCAN

dbscan = DBSCAN(eps=0.5, min_samples=5)
labels = dbscan.fit_predict(X)
```

**Silhouette Score**

 Metric to evaluate clustering performance

Measures how similar a point is to its own cluster vs others

Range: -1 to +1

+1: Well clustered

0: Borderline

-1: Wrong cluster

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X, labels)
print("Silhouette Score:", score)
```
