```python
# positional argument -- name
# keyword argument  -- age

def hell(name,age) :
  print("My name is {} and age is {}".format(name,age))
```

```python
hell("pavan",29)
```

```python
def hell(name,age=29) :
  print("My name is {} and age is {}".format(name,age))
```

```python
hell("pavan")
```

```python
def hell(*args,**kwargs) :
     print(args)
     print(kwargs)
```

```python
 hell('krish','naik',age=25,dob=1999)
```

```python
list=['krish', 'naik']
dict_args={'age': 25, 'dob': 1999}
```

```python
hell(list,dict_args)
```

```python
hell(*list,**dict_args)
```

#MAP


The map() function in Python is a built-in function that lets you apply a function to every item in an iterable (like a list, tuple, etc.) and returns a map object (which is an iterator). You can convert this map object into a list, set, or tuple.



##map(function, iterable)

#function: the function to apply to each item.

#iterable: the iterable whose items you want to transform.

```python
lst=[2,54,76,90,106]
```

```python
map(even_odd,lst)
```

```python
list(map(even_odd,lst))
```

#LAMBDA FUNCTION

#FILTER

```python
def even(num):
  if num%2==0:
    return True
```

```python
even(24)
```

```python
lst=[12,34,56,78,43]
```

```python
list(filter(even,lst))
```

```python
list(filter(lambda num:num%2==0,lst))
```

```python
list(map(lambda num:num%2==0,lst ))
```
