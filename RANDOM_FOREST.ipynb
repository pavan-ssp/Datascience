```python
from sklearn.tree import DecisionTreeClassifier, export_text

# Sample Data
X = [[22], [25], [28], [35], [40], [45], [50], [60], [65], [70]]  # Age
y = ["Yes", "No", "Yes", "No", "No", "Yes", "No", "Yes", "No", "Yes"]  # Buy Car?

# Train Decision Tree with Gini Impurity
clf = DecisionTreeClassifier(criterion="gini")
clf.fit(X, y)

# Print Decision Tree
tree_rules = export_text(clf, feature_names=["Age"])
print(tree_rules)
```

```python
from sklearn.tree import DecisionTreeClassifier

# Sample Data (Feature: Age, Label: Buy Car Yes/No)
X = [[22], [25], [28], [35], [40], [45], [50], [60], [65], [70]]
y = ["Yes", "No", "Yes", "No", "No", "Yes", "No", "Yes", "No", "Yes"]

# Train Decision Tree for Classification
clf = DecisionTreeClassifier(criterion="gini")
clf.fit(X, y)

# Make Prediction
print(clf.predict([[30]]))  # Predict if a 30-year-old will buy a car
```
