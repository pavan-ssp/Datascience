#Autograd

PyTorch makes **differentiation** (computing gradients) easy and efficient by providing the **autograd** package. This allows you to automatically compute the gradients of tensors during backpropagation, which is essential for training machine learning models, particularly in deep learning.


When you perform operations on tensors that have requires_grad=True, PyTorch builds a computation graph under the hood. When you call .backward(), it:

Traverses the graph

Applies the chain rule

Computes gradients automatically

```python
import torch

# Create a tensor with gradient tracking enabled
x = torch.tensor(2.0, requires_grad=True)

# Some operation
y = x ** 3    # y = x^3 = 8

# Backward pass
y.backward()

# Check gradient (dy/dx = 3x^2 = 3*4 = 12)
print(x.grad)   # tensor(12.)
```

#Functionality of Autograd

x.requires_grad=True tells PyTorch to track operations on x

y = x**3 creates a computation graph

y.backward() computes the gradient of y with respect to x

x.grad gives you the result: 12

## DataLoader and Dataset

#Dataset

Think of Dataset as a wrapper around your data. It helps PyTorch know:

What your inputs and targets are

How many samples you have

How to retrieve each sample

It's a class where you define how your dataset looks and how to access it.

How to Use:

You create a custom dataset by subclassing torch.utils.data.Dataset and implementing:

__len__() → returns number of samples

__getitem__(idx) → returns one data sample at index idx

#DataLoader

Batch your data (e.g., 32 samples at a time)

Shuffle it during training

Use multiple workers to speed up data loading

```python
from torch.utils.data import Dataset, DataLoader
```

```python
class CustomDataSet(Dataset):
    def __init__(self,features,labels):
        self.features=features
        self.labels=labels

    def __len__(self):
        return self.features.shape[0]

    def __getitem__(self,index):
        return self.features[index],self.labels[index]
```

```python
dataset=CustomDataSet(X,y)
```
